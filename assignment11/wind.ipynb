```python
import plotly.express as px
import plotly.data as pldata
df = pldata.wind(return_type='pandas')
display(df.head(10))
```

```python
df.info()
```

```python
df['strength'].value_counts()
```

```python
df['strength'] = df['strength'].replace({'6+': '6-6'})
df['strength'] = df['strength'].str.split('-').apply(lambda x: (int(x[0]) + int(x[1])) / 2)
df['strength'].head()
```

```python
fig = px.scatter(df, x='strength', y='frequency', color='direction',
                 title="Strength vs Frequency", hover_data=["strength"])
fig.write_html("wind.html", auto_open=True)
```
